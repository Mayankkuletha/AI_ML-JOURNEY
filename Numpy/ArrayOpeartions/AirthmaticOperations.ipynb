```python
import numpy as np

#we need two arrays for performing mathmatical operations

arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])

print(arr1)
print(arr2)
```

```python
# addition 
#In list we have to first create loops then do further things to perform opeartions
#But in numpy its just very simple
arr1+arr2
```

```python
# Substraction
arr1-arr2
```

```python
# Multiplication
arr1*arr2
```

```python
# Division
arr1/arr2
```

```python
# Floor division
arr1//arr2
```

```python
# Exponentional
arr1**arr2
```
