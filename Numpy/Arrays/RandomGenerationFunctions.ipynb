```python
import numpy as np

# Rand (print random numbers between 0 and 1 thats the concept of normalization)


a=np.random.rand(10)
print(a)
```

```python
# Randn --> print random number between -3 and 3 this is concept of standardization
a = np.random.randn(10)
print(a)
```

```python
# Randint ---> print random number in this we can give range and how many random numbers to print
a = np.random.randint(1,10,10)
print(a)
```
