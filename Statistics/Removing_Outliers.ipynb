Removal of Outliers and implementing box Plot

```python
import numpy as np
import pandas as pd
```

```python
arr1 = np.array([2,3,4,6,7,8,9,12,13,16,17,23,25,27,34,37,201])
arr1
```

```python
# Quartiles
Q1 = np.percentile(arr1,25)
Q3 = np.percentile(arr1,75)
IQR = Q3-Q1
```

```python
# Upper fence and lower fence
UF = Q3+(1.5*IQR)
LF = Q1 - (1.5*IQR)
```

```python
# removing outliers
l = []
for i in arr1:
    if i>=LF and i<=UF:
        l.append(i)

arr2 = np.array(l)
arr2
```

```python
# box plot
import seaborn as sns
sns.boxplot(x = arr2)
```
