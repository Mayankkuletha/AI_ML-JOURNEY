```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Distribution Plot
# we use distribution plot to visualize the distribution of quantative data.

# Lets discuss some plots that allow us to visualize the distribution of a data set . Thse plots are :

# * distplot/histplot
# * jointplot
# * pairplot
# * rugplot
```

```python
# seaborn has some inbuilt data sets we can also load them
df = sns.load_dataset("tips")
df
```

```python
# In above table total bill tip are quantative 
df['size'].unique()
# out of 243 only 6 are unique so we can also categorize it. so it is not quantitative
```

# Dist or Histplot ---> 
English:

histplot shows the distribution of a numerical variable using bins.

Hinglish:

histplot ek numerical variable ki values ko different ranges (bins) mein divide karke batata hai ki har range mein kitni values hain.

bins=5 → data ko 5 ranges mein divide karo.
kde(Kernel density estimate)=True → distribution ki smooth curve bhi add karta hai.
KDE ek smooth line/curve hoti hai jo batati hai ki data kis area mein zyada concentrated hai aur kis area mein kam

```python

plt.subplot(1,2,1)
sns.histplot(df['total_bill'] , bins=20 , kde=True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins=20,kde=True)


# in similar way we can use matplotlib and seaborn simantaneously.
```

Joint Plot --> English:

Joint plot shows the relationship between two numerical variables along with the individual distribution of each variable.

Hinglish:

Joint plot ek saath 2 variables ka relationship aur dono variables ki distribution dikhata hai.

```python
sns.jointplot(x='total_bill',y='tip',data=df,kind="scatter")
sns.jointplot(x='total_bill',y='tip',data=df,kind="kde")
sns.jointplot(x='total_bill',y='tip',data=df,kind="hex")
sns.jointplot(x='total_bill',y='tip',data=df,kind="reg")
```

PairPlot -->
English:

Pair plot shows the relationship between every pair of numerical variables in a dataset.

Hinglish:

Pair plot dataset ke multiple numerical columns ke har possible pair ka relationship ek saath dikhata hai.

1. hue
hue adds a categorical variable and uses different colors for its categories.
2. palette
palette controls which colors are used for the categories created by hue.


```python
# sns.pairplot(df) #will compare between diffrent numerical or quantative values.
sns.pairplot(df,hue='sex') #now it will compare with respect to sex (male,female) means we can categorize it.
sns.pairplot(df,hue='sex',palette="rainbow") 
```

Rugplot-->
English:

A rug plot shows the location of individual data points as small marks along an axis.

Hinglish:

Rug plot har individual data value ki position ko chhoti-chhoti lines (ticks) se show karta hai.

Har | = ek data point.

```python
sns.rugplot(df)
```
