```python
import matplotlib.pyplot as plt
import seaborn as sns
```

HeatMap

Heatmap ek matrix/table ke values ko colors ke through show karta hai.

Dark/strong color → value zyada
Light color → value kam

Sabse common use: Correlation dekhne ke liye.

```python
tips = sns.load_dataset("tips")
tips
```

```python
flight = sns.load_dataset("flights")
flight
```

```python
tipscorr=tips[['total_bill','tip','size']]
tipscorr
```

```python
tipscorr.corr()
```

```python
sns.heatmap(tipscorr.corr() , annot=True)
```

Cluster Map

clustermap() data ke rows aur columns ko similarity ke basis par automatically group (cluster) karta hai aur uske baad heatmap banata hai.


Heatmap: bas values ko colors mein dikhata hai.
Cluster Map: values ko colors mein dikhata hai + similar rows/columns ko paas-paas group karta hai.



Bhai annot ka matlab annotation hota hai — yani graph ke andar actual values/numbers likh dena.

```python
sns.clustermap(tipscorr.corr(), annot = True)
```

HeatMap Using Pivot Table

```python
fl = flight.pivot_table(index='month' , values='passengers' , columns='year')
fl
```

```python
sns.heatmap(fl)
```
