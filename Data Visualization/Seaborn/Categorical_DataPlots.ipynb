```python
import matplotlib.pyplot as plt
import seaborn as sns
```

Categorical Data Plots

Now lets discuss using seaborn to plot categorical data. There are few main plot types for this :

* boxPlot
* violinPlot
* stripplot
* swarmplot
* barplot
* countplot

```python
df = sns.load_dataset('tips')
df
```

CountPlot 

English:

countplot shows the number of observations in each category.

Hinglish:

Countplot har category mein kitni values/entries hain, woh count karke bars mein dikhata hai

countplot mein hue ka main kaam grouping karna hai.

```python
# sns.countplot(df['sex']) #by default it will show for y axis you can make it x axis also
sns.countplot(x=df['sex'] , hue= df['smoker'])
```

BarPlot

English:

Barplot compares a numerical value across different categories using bars.

Hinglish:

Barplot different categories ke numerical values ko compare karta hai.

```python
sns.barplot(x=df['sex'] , y=df['total_bill'])
```

```python
# You can change the estimator object to your own function , that converts a vector to a scalar Mean (default) : np.mean Median :np.median 
#sum : np.sum Count: len(to count the number of observations)  Standard Deviation : np.std Minimum : np.min  Maximum : np.max
import numpy as np
sns.barplot(x=df['sex'] , y=df['total_bill'] , estimator=np.sum)
sns.barplot(x=df['sex'] , y=df['tip'] , estimator=np.sum)
```

Boxplot

English:

A boxplot summarizes the distribution of numerical data using median, quartiles, spread, and outliers.

Hinglish:

Boxplot numerical data ka spread, median aur outliers ko ek graph mein show karta hai.

```python
sns.boxplot(x=df['tip'] , y= df['day'] ,data=df, palette='rainbow')
```

Violin plot

English:

Violin plot shows the distribution of numerical data, combining a boxplot with a KDE (density) plot.

Hinglish:

Violin plot boxplot + KDE dono ko combine karke data ka distribution dikhata hai.

```python
sns.violinplot(x='tip' , y='day' , data=df)
```

```python
# boxplots and violinplots are used to show the distribution of categorical data. A box plot (or box and whisker plot) shows the distribution of quantitative data in a way that facilitates comparisons between variables or across levels of a categorical variable . The box shows
# the quartiles of the dataset while the whiskers extend to show the rest of the distribution , except for points that are determined to be "outliers" using a method that is a function of the inter-quartile range 
```

Strip Plot

English:

A strip plot shows individual data points for each category as dots.

Hinglish:

Strip plot har category ke individual data points ko dots ke form mein dikhata hai.

```python
sns.stripplot(x='day' , y='tip' , data= df  )
```

Swarm plot

English:

Swarm plot shows individual data points for each category without overlapping them.

Hinglish:

Swarm plot strip plot jaisa hota hai, lekin ye dots ko automatically idhar-udhar arrange karta hai taaki points ek doosre ke upar hide na hon.

```python
sns.swarmplot(x='day', y= 'tip' , data=df)
```

```python
# we can also combine both swarm and violin plot as they were looking similar.
sns.swarmplot(x='tip' , y='day' , data=df )
sns.violinplot(x = 'tip' , y='day' , data= df )
```
