```python
import matplotlib.pyplot as plt
import seaborn as sns
```

lmport()  -> using this we create regression plots

Regression plot do numerical variables ke beech relationship dikhata hai aur unke beech ek best-fit line draw karta hai.

```python
tips = sns.load_dataset("tips")
tips
```

```python
sns.lmplot(x = "total_bill" , y ='tip' , data=tips)
```

```python
sns.lmplot(x = 'total_bill' , y='tip' ,data=tips, hue="sex")
```

```python
sns.lmplot(x = 'total_bill' , y='tip' ,data=tips, hue="sex" , palette="coolwarm")
```

```python
sns.lmplot(x = 'total_bill' , y='tip' ,data=tips, hue="sex" , palette="coolwarm",markers=['o','v'])
```

```python
sns.lmplot(x='total_bill',y='tip' , data=tips , hue="sex" , palette="coolwarm" , markers=['o','v'],
           scatter_kws={"s":100})
```
