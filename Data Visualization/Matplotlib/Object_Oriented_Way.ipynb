```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
y = x*x
y
```

```python
# Matplotlib allows the aspect ratio , DPI and figure size to be specified when the Figure Object is created .
#You can use the figsize and dpi keyword arguments.

# figsize is a tuple of the width and height of the figure in inches.
# dpi is the dots per inch (pixels per inch).  100 is default
# dpi means dots per inch and controls the resolution/detail of the Figure.
```

```python
fig = plt.figure(figsize=(14,7),dpi=100)  #creates a figure or empty space to draw graph
axis1 = fig.add_axes([0.1,0.1,0.8,0.8])
axis1.plot(x,y)
axis2 = fig.add_axes([0.2,0.4,0.2,0.4])
axis2.plot(x,y)
```
