```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Woking with Images

import matplotlib.image as mpimg
img =mpimg.imread("workingwithimage.jpg")
# img #--> shows the array of pixels of images. means it convert it into array based on pixels of images.

```

```python
plt.imshow(img)
plt.axis("off") #will remove axis/
```

```python
# we can also crop (we will access that array of pixels of img)
crop_img = img[50:200,100:300] #crop --> rows 50-200 , colms 100,300
plt.imshow(crop_img)
plt.axis("off")
```
