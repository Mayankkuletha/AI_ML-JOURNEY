```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
y = x**2
y
```

```python
# Scatter Plot -->Scatter plot shows the relationship between two numerical variables using individual points.

# Hinglish:

# Scatter plot mein do numerical variables ke beech relationship ko dots/points ke through dikhate hain.

plt.scatter(x,y)
```

```python
# Histogram 
# English:

# Histogram shows the distribution of numerical data by dividing values into ranges called bins.

# Hinglish:

# Histogram batata hai ki data ki values kis range mein kitni baar aa rahi hain.
plt.hist(x)

# takes only one value other automatically works with points
```

```python
# boxplot --> follows fine number summary

# English:

# Boxplot summarizes the distribution of numerical data using median, quartiles, spread, and outliers.

# Hinglish:

# Boxplot data ka distribution, median, spread aur outliers ek compact graph mein dikhata hai.
plt.boxplot(x)
```
