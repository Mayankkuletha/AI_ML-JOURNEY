```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
y = x**2
y
```

```python
# Subplots 
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x**2)
plt.subplot(2,2,4)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x**2,x**2)
```

```python
# Subplot -->Subplot means creating multiple graphs inside a single Figure.
# Figure
# ┌─────────────┬─────────────┐
# │   Graph 1   │   Graph 2   │
# ├─────────────┼─────────────┤
# │   Graph 3   │   Graph 4   │
# └─────────────┴─────────────┘
```
