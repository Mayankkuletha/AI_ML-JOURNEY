```python
import numpy as np
import pandas as pd

employee = pd.DataFrame({
    "Employee_ID": [101, 102, 103, 104],
    "Name": ["Mayank", "Himanshu", "Karan", "Rahul"],
    "Department": ["IT", "HR", "Finance", "IT"]
})

salary = pd.DataFrame({
    "Employee_ID": [103, 104, 105, 106],
    "Salary": [55000, 70000, 45000, 60000]
})

employee
```

```python
salary
```

```python
# Merging both

pd.merge(employee,salary , on="Employee_ID") #we should also give on parameter to tell him on which column basic we have to merge
```

```python
pd.merge(employee,salary,on="Employee_ID",how="inner")

#BY default how is always set to inner just give common between both
```

```python
pd.merge(employee,salary,on="Employee_ID",how="outer")

# outer --> It will give both tables value and the unknown value will be filled with nan
```

```python
pd.merge(employee,salary,on="Employee_ID",how="left")

# left --> Will give me the values (Employee_Id)  based on the employee table ,all Emp_id will be there  
```

```python
pd.merge(employee,salary,on="Employee_ID",how="right")
```

```python
# right --> Will give me the values (Employee_Id)  based on the salary table ,all Emp_id will be there .
```
