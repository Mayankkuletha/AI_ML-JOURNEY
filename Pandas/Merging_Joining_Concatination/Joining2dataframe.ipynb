```python
# Joining Two data frames

import numpy as np
import pandas as pd



df1 = pd.DataFrame({
    "Name": ["Mayank", "Himanshu", "Karan", "Rahul"]
}, index=[1, 2, 3, 4])

df2 = pd.DataFrame({
    "Score": [78, 92, 85, 88]
}, index=[3, 4, 5, 6])


df1
```

```python
df2
```

```python
# BY default join (how is always inner)

df1.join(df2 , how='inner')

# if df2.join(df1.join) was written then score will come ahead in the table.
```

```python
# Outer join 

df1.join(df2,how='outer')
```
