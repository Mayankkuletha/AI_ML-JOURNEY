```python
import numpy as np
import pandas as pd

data = {
    "Name": ["Mayank", "Himanshu", "Karan", "Rahul", "Aman", "Vikas"],
    "Age": [20, 22, np.nan, 25, 21, 18],
    "City": ["Delhi", "Mumbai", "Pune", np.nan, "Jaipur", "Dehradun"],
    "Salary": [50000, 60000, 45000, 70000, np.nan, 40000]
}

df = pd.DataFrame(data)
df
```

```python
# If you want to remove whole row if it contain null value

df.dropna()
```

```python
# if you want to remove only that row which atleast some non null values

df.dropna(thresh=4) #Should contain atleast 4 non null values.
```
