```python
import numpy as np
import pandas as pd

data = {
    "Name": ["Mayank", "Himanshu", "Karan", "Rahul", "Aman", "Vikas"],
    "Age": [20, 22, np.nan, 25, 21, 18],
    "City": ["Delhi", "Mumbai", "Pune", np.nan, "Jaipur", "Dehradun"],
    "Salary": [50000, 60000, 45000, 70000, np.nan, 40000]
}

df = pd.DataFrame(data)
df
```

```python
#to check whether Nan values is there in it 

df.isna() #wherever nan will be there it will give true otherwise false
```

```python
# total nan in each column.
df.isna().sum()
```

```python
# to check whether each column  contain nan value the column with no nan value will give false.

df.isna().any()
```
