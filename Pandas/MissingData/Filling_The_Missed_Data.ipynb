```python
import numpy as np
import pandas as pd

data = {
    "Name": ["Mayank", "Himanshu", "Karan", "Rahul", "Aman", "Vikas"],
    "Age": [20, 22, np.nan, 25, 21, 18],
    "City": ["Delhi", "Mumbai", "Pune", np.nan, "Jaipur", "Dehradun"],
    "Salary": [50000, 60000, 45000, 70000, np.nan, 40000]
}

df = pd.DataFrame(data)
df
```

```python
# to fill all nan with specific values

df.fillna(0)  

#NOTE --> this will not change it in original datauntil we write inplace = true.
```

```python
#to fill with our given values

values = {"Name" : "Nothing"  , "Age": 20, "City":"UNKNOWN" , "Salary":1}
df.fillna(value=values)
```
