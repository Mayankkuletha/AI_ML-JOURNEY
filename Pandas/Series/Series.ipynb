```python
import numpy as np
import pandas as pd

labels = ['a','b','c']
my_list=[10,20,30]
arr = np.array([10,20,30])
d = {'a':10,'b':20,'c':30}
```

```python
# Making series of list
pd.Series(my_list)
pd.Series(my_list,index = labels) #It will change the indexes from 0 ,1 , 2 to values of labels

# a,b,c are labels and 10 20 30 are data points
```

```python
# Making Series of Arrays
pd.Series(arr)
pd.Series(arr,index = labels)

# We cannot pass 2D array because Series itself is one Dimensional
```

```python
# Making Series from Dictionary
pd.Series(d)
# Keys act as labels here
```
